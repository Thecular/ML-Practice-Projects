```python
'''Area to get good at in data cleaning are finding areas of null values(.first_valid_index()),delecting(dropna,drop) corresponding null-value rows.'''
#titanic['Age'] = titanic['Age']#How do I change the age to only integer.
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].mean(),inplace=False)
```

```python
titanic.drop(columns=['Ticket'],inplace=True)# Dropped ticket and cabin column.
```

```python
titanic['Embarked'] = titanic['Embarked'].fillna(titanic['Embarked'].mode())
```
